```python
# ATIVIDADE - Bloco 1

produtos = ["Camiseta Basica","Calça Jeans","Jaqueta Corta-Vento"]
precos = [49.90,129.90,219.90]

for i,j in zip(produtos, precos):
  print(f"o produto {i} custa {j}")
```

```python
# ATIVIDADE - Bloco 2

novo_produto = input("Informe o nome do novo produto: ")
novo_preco = float(input("Informe o preço do novo produto: "))
print("Produto cadastrado com sucesso!")

produtos.append(novo_produto)
precos.append(novo_preco)
for i,j in zip(produtos,precos):
  print(f"A {i} custa R${j}")
```
